## Dataset statistics

```
import json
import pandas as pd

stats = json.load(open("data/processed/stats.json"))
df = pd.DataFrame(stats).T
df.index = pd.MultiIndex.from_tuples([tuple(k.split("/")) for k in df.index], names=["domain", "split"])
df["neutral_%"] = (100 * df["NEU"] / df["quads"]).round(1)
df["implicit_aspect_%"] = (100 * df["implicit_aspect"] / df["quads"]).round(1)
df[["sentences", "quads", "NEG", "NEU", "POS", "neutral_%", "implicit_aspect_%", "conflicting_sentences", "categories", "mean_words"]]
```

```
totals = df.groupby(level="split")[["sentences", "quads", "NEG", "NEU", "POS", "conflicting_sentences"]].sum()
totals.loc["All"] = totals.sum()
totals
```
